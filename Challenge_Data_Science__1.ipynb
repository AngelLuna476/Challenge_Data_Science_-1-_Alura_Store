### Primera tienda

```python

# Cargar el archivo .csv
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"

data = pd.read_csv(url)
data['Lugar de Compra'] = data['Lugar de Compra'].astype(str)


# Gráfico de dispersión para la distribución geográfica de las ventas
plt.figure(figsize=(10, 6))
sns.scatterplot(x=data['lon'], y=data['lat'], size=data['Lugar de Compra'], hue=data['Lugar de Compra'], palette='viridis')
plt.title('Distribución geográfica de las ventas')
plt.xlabel('Longitud')
plt.ylabel('Latitud')
plt.legend(title='Distribución de ventas')
plt.show()

# Crear un mapa de calor (Heatmap) con Folium
# Crear el mapa base centrado en la media de las coordenadas
m = folium.Map(location=[data['lat'].mean(), data['lon'].mean()], zoom_start=10)


# Agregar los datos al Heatmap
heat_data = [[row['lat'], row['lon']] for index, row in data.iterrows() if pd.notna(row['lat']) and pd.notna(row['lon'])]  # Filter out NaN values

# Crea the HeatMap object
HeatMap(heat_data).add_to(m)


# Guardar y mostrar el mapa
m.save('mapa_calor1.html')  # Guarda el mapa en un archivo HTML
print("El mapa de calor se guardó como 'mapa_calor1.html'")



# Cargar los datos desde la URL
data = pd.read_csv(
    url,
    delimiter=',',       # delimitador si es necesario
    encoding='utf-8',    # codificación si es necesario
    quotechar='"'        # comillas si es necesario
)

# Contar las ocurrencias de cada producto
conteo_desempeno_tienda_1 = data['Lugar de Compra'].value_counts().reset_index()
conteo_desempeno_tienda_1.columns = ['Lugar de Compra', 'Ocurrencias']

# Ordenar por la cantidad de ocurrencias en orden descendente
conteo_desempeno_tienda_1_ordenado = conteo_desempeno_tienda_1.sort_values('Ocurrencias', ascending=False)

# Imprimir los resultados en la consola
print("Conteo de los Lugares de Compra de la primera tienda:")

# Mostrar los datos como una tabla interactiva
from IPython.display import display
display(conteo_desempeno_tienda_1_ordenado)  # Muestra las categorías de mayor a menor y su conteo
```
